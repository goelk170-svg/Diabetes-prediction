```python
df['Gender'] = df['Gender'].map({'Female': 0,'Male': 1,'Unknown': 2})
df['Gender'].unique()
```

```python
df['Smoking_Status'] = df['Smoking_Status'].map({'Never': 0,'Former': 1,'Current': 2})
df['Smoking_Status'].unique()
```

```python
df['Physical_Activity'] = df['Physical_Activity'].map({'Low': 0,'Moderate': 1,'High': 2})
df['Physical_Activity'].unique()
```

```python
df['Family_History'] = df['Family_History'].map({'No': 0,'Yes': 1})
df['Family_History'].unique()
```

```python
df['Diabetes'].unique()
```

```python
df['Diabetes'] = df['Diabetes'].map({'No': 0,'Yes': 1})
df['Diabetes'].unique()
```

```python
df.head()
```

```python
(df.isnull().sum())
```

```python
df['BMI'] = df['BMI'].fillna(df['BMI'].median())
df['Glucose'] = df['Glucose'].fillna(df['Glucose'].median())
```

```python
df.isnull().sum()
```

```python
X = df.drop(['Patient_ID', 'Diabetes'], axis=1)
y = df['Diabetes']

X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn import metrics
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
X
```

```python
X_test.head()
```

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)

rf = RandomForestClassifier()
rf.fit(X_train, y_train)

xgb = GradientBoostingClassifier()
xgb.fit(X_train, y_train)

xg = XGBClassifier()
xg.fit(X_train, y_train)
```

```python
y_pred1 = lr.predict(X_test)
y_pred2 = rf.predict(X_test)
y_pred3 = xgb.predict(X_test)
y_pred4 = xg.predict(X_test)
```

```python
y_test.head()
```

```python
y_pred1
```

```python
accuracy1 = metrics.accuracy_score(y_test, y_pred1)
accuracy2 = metrics.accuracy_score(y_test, y_pred2)
accuracy3 = metrics.accuracy_score(y_test, y_pred3)
accuracy4 = metrics.accuracy_score(y_test, y_pred4)
```

```python
f1_1 = metrics.f1_score(y_test, y_pred1)
f1_2 = metrics.f1_score(y_test, y_pred2)
f1_3 = metrics.f1_score(y_test, y_pred3)
f1_4 = metrics.f1_score(y_test, y_pred4)
```

```python
precision1 = metrics.precision_score(y_test, y_pred1)
precision2 = metrics.precision_score(y_test, y_pred2)
precision3 = metrics.precision_score(y_test, y_pred3)
precision4 = metrics.precision_score(y_test, y_pred4)
```

```python
recall1 = metrics.recall_score(y_test, y_pred1)
recall2 = metrics.recall_score(y_test, y_pred2)
recall3 = metrics.recall_score(y_test, y_pred3)
recall4 = metrics.recall_score(y_test, y_pred4)
```

```python
final_data = pd.DataFrame({
    'Models': ['LR', 'RF', 'GB', 'XG'],
    'Accuracy': [accuracy1, accuracy2, accuracy3, accuracy4],
    'Precision': [precision1, precision2, precision3, precision4],
    'Recall': [recall1, recall2, recall3, recall4],
    'F1_SCORE': [f1_1, f1_2, f1_3, f1_4]
})
```

```python
final_data.index = range(1, len(final_data) + 1)
final_data
```

```python
final_data_melted = final_data.melt(id_vars='Models',value_vars=['Accuracy', 'F1_SCORE'],var_name='Metrics',value_name='Score')
sns.barplot(x='Models',y='Score',hue='Metrics',data=final_data_melted)
plt.title('Accuracy vs F1 Score Comparison')
plt.xlabel('Models')
plt.ylabel('Score')
plt.legend(title='Metrics',  bbox_to_anchor=(1, 0.5))
plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
models = [('Logistic Regression', y_pred1),('Random Forest', y_pred2),('Gradient Boosting', y_pred3),('XGBoost', y_pred4)]

for ax, (name, prediction) in zip(axes.flatten(), models):
    ConfusionMatrixDisplay.from_predictions(y_test,prediction,display_labels=['No Diabetes', 'Diabetes'],cmap='Blues',ax=ax,colorbar=False,
                                            values_format='d')
    ax.set_title(name)
plt.suptitle('Confusion Matrix Comparison of Four Models')
plt.tight_layout()
plt.show()
```
