```python
import pandas as pd
```

```python
df=pd.read_excel('Superstore (1).xlsx')
df.head()
                 
```

```python
category_sales=df.groupby('Category')['Sales'].sum()
plt.figure(figsize=(4,3))
category_sales.plot(kind='bar',color='green')
plt.xticks(rotation=45)
plt.show()
```

```python
region_profit=df.groupby('Region')['Profit'].sum()
plt.figure(figsize=(4,3))
region_profit.plot(kind='barh',color='brown')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales=df.groupby('Category')['Sales'].sum()
plt.pie(category_sales, labels=category_sales.index, autopct='%1.1f%%', startangle=100)
plt.xticks(rotation=45)
plt.show()
```

```python
subcat=df.groupby("Sub-Category")[['Sales','Profit']].sum().reset_index()
plt.scatter(subcat['Sales'], subcat['Profit'])
for i in range(len(subcat)):
    plt.text(subcat['Sales'][i], subcat['Profit'][i], subcat["Sub-Category"][i])
    print(subcat['Sales'][i])
plt.xlabel("total Sales")
plt.ylabel("total Profit")
plt.title("Sales vs Profit by Sub-Category")
plt.show()
```

```python
x1 = np.array([10,20,30,40])
y1 = np.array([10,20,30,40])
plt.subplot(1,2,1)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product & Sales")
plt.plot(x1,y1)

x2 = np.array([10,50,30,80])
y2 = np.array([10,20,30,40])
plt.subplot(1,2,2)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product & Profit")
plt.plot(x2,y2)

plt.suptitle("Products wise Sales and Profit")
plt.tight_layout()
plt.show()
```
